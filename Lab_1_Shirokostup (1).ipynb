```python
import math

# --- Завдання 1.2: Кусочно-задана функція та цикл на сітці ---
a_12 = 0.75
b_12 = 1.19
c_12 = -2.5
L = 0.0
R = 1.0
h = 0.1

print("і |  x  |     гілка     |      y")
print("--|-----|---------------|-------------------")

m = int((R - L) / h)

for i in range(m + 1):
    x = round(L + i * h, 10)
    
    if 0 <= x < 0.5:
        branch = "1-а [0; 0.5)"
        y = a_12 * x + b_12 * math.cos(x)
    elif 0.5 <= x <= 1.0:
        branch = "2-а [0.5; 1]"
        y = b_12 * (x ** 2) + c_12 * math.sin(2 * x)
    else:
        branch = "поза межами"
        y = None
        
    if y is not None:
        print(f"{i:2d} | {x:3.1f} | {branch} | {y:11.5f}")
    else:
        print(f"{i:2d} | {x:3.1f} | {branch} | —")
```
